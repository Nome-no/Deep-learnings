```python
# 1. Función de la Neurona
def perceptron_manual(entradas, pesos, sesgo):
    suma_ponderada = 0

    # Recorremos cada posición de la lista de entradas usando su índice (0, 1, 2...)
    # len(entradas) nos dice cuántos elementos hay
    for i in range(len(entradas)):
        valor_x = entradas[i]    # Tomamos la entrada en la posición i
        valor_w = pesos[i]       # Tomamos el peso en la misma posición i

        producto = valor_x * valor_w  # Multiplicamos
        suma_ponderada = suma_ponderada + producto  # Vamos acumulando el resultado

    # Al final de la suma de todos los productos, sumamos el sesgo
    z = suma_ponderada + sesgo

    # Regla de activación (Umbral)
    if z >= 0:
        salida = 1
    else:
        salida = 0

    return z, salida

# 2. Configuración (Compuerta AND)
pesos_and = [0.6, 0.6]
sesgo_and = -1.0
casos_prueba = [[0,0], [0,1], [1,0], [1,1]]

# 3. Lista para guardar resultados
historial = []

# 4. Ejecución detallada con APPEND
for prueba in casos_prueba:
    # Llamamos a nuestra función
    puntaje_z, prediccion = perceptron_manual(prueba, pesos_and, sesgo_and)

    # Creamos una "fila" de datos manualmente
    fila_de_datos = [prueba, puntaje_z, prediccion]

    # APPEND: Metemos esa fila al final de nuestra lista 'historial'
    historial.append(fila_de_datos)

# 5. Mostrar resultados
print("Entrada | Z (Puntaje) | Clasificación")
print("-" * 40)
for fila in historial:
    print(f"{fila[0]}  | {fila[1]:>11} | {fila[2]}")
```
